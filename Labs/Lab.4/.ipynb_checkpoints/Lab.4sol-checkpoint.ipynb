```python
class counter:
    cnt = 0

    def __init__(self, max):
        self.max = max
    def increment(self):
        if(counter.cnt < self.max):
            counter.cnt += 1
            print(counter.cnt)
        else:
            print("Error: Counter has reached the maximum value.")
    def reset(self):
        counter.cnt = 0
        print("Counter has been reset.")
        
c1 = counter(3)
c1.increment()  
c1.increment()
c1.increment()  
c1.increment()  
c1.reset()      
c1.increment()  
```

2. Copy and paste your solution to question 1 and modify it so that all the data held by the counter is private. Implement functions to check the value of the counter, check the maximum value, and check if the counter is at the maximum.

```python
class counter:
    __cnt = 0  # Private counter
    def __init__(self, max):
        self.__max = max  # Private max value
    def increment(self):
        if counter.__cnt < self.__max:
            counter.__cnt += 1
            print(counter.__cnt)
        else:
            print("Error: Counter has reached the maximum value.")
    def reset(self):
        counter.__cnt = 0
        print("Counter has been reset.")
    def get_value(self):
        return counter.__cnt  
    def get_max(self):
        return self.__max  
    def is_at_max(self):
        return counter.__cnt == self.__max  

c1 = counter(3)
c1.increment()  
c1.increment()  
c1.increment()  
c1.increment()  

print("Current value:", c1.get_value())  
print("Maximum value:", c1.get_max())    
print("Is at max?", c1.is_at_max())      

c1.reset()      
print("After reset:", c1.get_value())  
```

3. Implement a class to represent a rectangle, holding the length, width, and $x$ and $y$ coordinates of a corner of the object. Implement functions that compute the area and perimeter of the rectangle. Make all data members private and privide accessors to retrieve values of data members. 

```python
class rectangle:
    def __init__(self, x, y, wd, ln):
        self.__x = x
        self.__y = y
        self.__length = ln
        self.__width = wd 

    def get_coordinates(self):
        return (self.__x, self.__y)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)


r1 = rectangle(5, 10, 7, 8)  

print("Coordinates of corner:", r1.get_coordinates())
print("Length:", r1.get_length())
print("Width:", r1.get_width())
print("Area is", r1.area())
print("Perimeter is", r1.perimeter())
```

```python
class circle:
    def __init__(self, x, y, r):
        self.__x = x
        self.__y = y
        self.__radius = r
    def get_center(self):
        return (self.__x, self.__y)
    def get_radius(self):
        return self.__radius
    def area(self):
        return 3.14 * self.__radius * self.__radius
    def perimeter(self):
        return 2 * 3.14 * self.__radius

c1 = circle(6, 9, 5)

print("Area is", c1.area())
print("Perimeter is", c1.perimeter())
print("Center is", c1.get_center())
print("Radius is", c1.get_radius())
```

6. Implement a triangle class analogous to the rectangle and circle in question 5.

```python
import math

class Base:
    pass  

class Triangle(Base):
    def __init__(self, x1, y1, x2, y2, x3, y3, a=None, b=None, c=None):
        self._x1 = x1
        self._y1 = y1
        self._x2 = x2
        self._y2 = y2
        self._x3 = x3
        self._y3 = y3

        self._a = a if a is not None else self._distance(x1, y1, x2, y2)
        self._b = b if b is not None else self._distance(x2, y2, x3, y3)
        self._c = c if c is not None else self._distance(x3, y3, x1, y1)

    def _distance(self, x1, y1, x2, y2):
        return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

    def area(self):
        # Using Heron's Formula
        perimeter = self._a + self._b + self._c
        s = perimeter / 2
        ar = math.sqrt(s * (s - self._a) * (s - self._b) * (s - self._c))
        print("Area is", ar)
        return ar

    def perimeter(self):
        perimeter = self._a + self._b + self._c
        print("Perimeter is", perimeter)
        return perimeter

    def __repr__(self):
        return f"Triangle({self._x1}, {self._y1}, {self._x2}, {self._y2}, {self._x3}, {self._y3}, {self._a}, {self._b}, {self._c})"

tr = Triangle(10, 10, 20, 20, 5, 20)
tr.area()
tr.perimeter()
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the parameter of the object. 

```python
import math

class Base:
    def get_perimeter_points(self):
        raise NotImplementedError("This method should be overridden by subclasses.")

class Rectangle(Base):
    def __init__(self, x, y, width, length):
        self.x = x
        self.y = y
        self._width = width
        self._length = length
    def get_perimeter_points(self):
        points = []
        step_width = self._width / 4
        step_length = self._length / 4

        for i in range(5):
            points.append((self.x + i * step_width, self.y))

        for i in range(1, 5):
            points.append((self.x + self._width, self.y + i * step_length))

        for i in range(1, 5):
            points.append((self.x + self._width - i * step_width, self.y + self._length))

        for i in range(1, 4):
            points.append((self.x, self.y + self._length - i * step_length))

        return points[:16]

class Circle(Base):
    def __init__(self, x, y, radius):
        self.x = x
        self.y = y
        self._radius = radius

    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = (2 * math.pi / 16) * i
            px = self.x + self._radius * math.cos(angle)
            py = self.y + self._radius * math.sin(angle)
            points.append((px, py))
        return points

class Triangle(Base):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self._x1 = x1
        self._y1 = y1
        self._x2 = x2
        self._y2 = y2
        self._x3 = x3
        self._y3 = y3

    def get_perimeter_points(self):
        points = []

        def get_points_between(x_start, y_start, x_end, y_end, n=5):
            return [
                (x_start + i * (x_end - x_start) / n, y_start + i * (y_end - y_start) / n)
                for i in range(n)
            ]

        points += get_points_between(self._x1, self._y1, self._x2, self._y2)
        points += get_points_between(self._x2, self._y2, self._x3, self._y3)
        points += get_points_between(self._x3, self._y3, self._x1, self._y1)

        return points[:16]

class Objects:
    def check_inside_rectangle(self, r1, x1, y1):
        return (r1.x < x1 < r1.x + r1._width) and (r1.y < y1 < r1.y + r1._length)
    def check_inside_circle(self, c1, x1, y1):
        return (x1 - c1.x) ** 2 + (y1 - c1.y) ** 2 < c1._radius ** 2
    def check_inside_triangle(self, t1, xp, yp):
        c1 = (t1._x2 - t1._x1) * (yp - t1._y1) - (t1._y2 - t1._y1) * (xp - t1._x1)
        c2 = (t1._x3 - t1._x2) * (yp - t1._y2) - (t1._y3 - t1._y2) * (xp - t1._x2)
        c3 = (t1._x1 - t1._x3) * (yp - t1._y3) - (t1._y1 - t1._y3) * (xp - t1._x3)
        return (c1 < 0 and c2 < 0 and c3 < 0) or (c1 > 0 and c2 > 0 and c3 > 0)


r1 = Rectangle(5, 5, 80, 60)         
c1 = Circle(50, 50, 30)              
t1 = Triangle(10, 10, 60, 20, 30, 70)  
o1 = Objects()

print("=== Rectangle Check ===")
print("Point is inside" if o1.check_inside_rectangle(r1, 20, 20) else "Point is outside")
print("Point is inside" if o1.check_inside_rectangle(r1, 100, 100) else "Point is outside")

print("\n=== Circle Check ===")
print("Point is inside" if o1.check_inside_circle(c1, 40, 40) else "Point is outside")
print("Point is inside" if o1.check_inside_circle(c1, 100, 100) else "Point is outside")

print("\n=== Triangle Check ===")
print("Point is inside" if o1.check_inside_triangle(t1, 30, 40) else "Point is outside")
print("Point is inside" if o1.check_inside_triangle(t1, 100, 100) else "Point is outside")

print("\n=== Perimeter Points ===")
print("Rectangle Perimeter Points:", r1.get_perimeter_points())
print("Circle Perimeter Points:", c1.get_perimeter_points())
print("Triangle Perimeter Points:", t1.get_perimeter_points())
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
import math

class Shape:
    def check_inside(self, x, y):
        raise NotImplementedError("This method should be overridden by subclasses.")

    def get_16_pair(self):
        lst = []
        for i in range(1, 17):
            x1 = i
            y1 = i
            lst.append((x1, y1))
        return lst

class Rectangle(Shape):
    def __init__(self, x, y, width, length):
        self.x = x
        self.y = y
        self._width = width
        self._length = length
    def check_inside(self, x1, y1):
        return (self.x < x1 < self.x + self._width) and (self.y < y1 < self.y + self._length)

class Circle(Shape):
    def __init__(self, x, y, radius):
        self.x = x
        self.y = y
        self._radius = radius
    def check_inside(self, x1, y1):
        return (x1 - self.x) ** 2 + (y1 - self.y) ** 2 < self._radius ** 2

class Triangle(Shape):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self._x1 = x1
        self._y1 = y1
        self._x2 = x2
        self._y2 = y2
        self._x3 = x3
        self._y3 = y3
    def check_inside(self, xp, yp):
        # Barycentric coordinates approach
        def sign(x1, y1, x2, y2, x3, y3):
            return (x1 - x3) * (y2 - y3) - (x2 - x3) * (y1 - y3)

        d1 = sign(xp, yp, self._x1, self._y1, self._x2, self._y2)
        d2 = sign(xp, yp, self._x2, self._y2, self._x3, self._y3)
        d3 = sign(xp, yp, self._x3, self._y3, self._x1, self._y1)

        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)

        return not (has_neg and has_pos)

rect = Rectangle(10, 10, 100, 100)
circle = Circle(50, 50, 30)
triangle = Triangle(0, 0, 50, 0, 25, 50)

points_to_test = rect.get_16_pair()

print("=== Rectangle Check ===")
for x, y in points_to_test:
    result = rect.check_inside(x, y)
    print(f"Point ({x}, {y}) is {'inside' if result else 'outside'} the rectangle.", end=' ')

print("\n\n=== Circle Check ===")
for x, y in points_to_test:
    result = circle.check_inside(x, y)
    print(f"Point ({x}, {y}) is {'inside' if result else 'outside'} the circle.", end=' ')

print("\n\n=== Triangle Check ===")
for x, y in points_to_test:
    result = triangle.check_inside(x, y)
    print(f"Point ({x}, {y}) is {'inside' if result else 'outside'} the triangle.", end=' ')
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python
import math

class Base(object):
    def area(self):
        raise NotImplementedError()
    def perimeter(self):
        raise NotImplementedError()
    def check_overlap(self, other):
        raise NotImplementedError()


class Circle(Base):
    def __init__(self, x, y, r):
        self.x = x
        self.y = y
        self._radius = r
    def area(self):
        ar = math.pi * self._radius ** 2
        print("Area is", ar)
    def perimeter(self):
        pr = 2 * math.pi * self._radius
        print("Perimeter is", pr)
    def check_overlap(self, other):
        if isinstance(other, Circle):
            dist_sq = (self.x - other.x) ** 2 + (self.y - other.y) ** 2
            rad_sum_sq = (self._radius + other._radius) ** 2
            return dist_sq <= rad_sum_sq
        return False


class Rectangle(Base):
    def __init__(self, x, y, width, length):
        self.x = x
        self.y = y
        self._width = width
        self._length = length
    def area(self):
        ar = self._width * self._length
        print("Area is", ar)
    def perimeter(self):
        pr = 2 * (self._width + self._length)
        print("Perimeter is", pr)
    def check_overlap(self, other):
        if isinstance(other, Rectangle):
            # Check if rectangles do not overlap
            if (self.x + self._width <= other.x or other.x + other._width <= self.x or
                    self.y + self._length <= other.y or other.y + other._length <= self.y):
                return False
            return True
        return False

r1 = Rectangle(2, 2, 3, 4)
r2 = Rectangle(3, 3, 3, 4)

print("=== Rectangle Overlap Check ===")
if r1.check_overlap(r2):
    print("Rectangles Overlap")
else:
    print("Rectangles Don't Overlap")

c1 = Circle(3, 4, 10)
c2 = Circle(5, 5, 10)

print("\n=== Circle Overlap Check ===")
if c1.check_overlap(c2):
    print("Circles Overlap")
else:
    print("Circles Don't Overlap")
```

10. Copy the `Canvas` class from lecture to in a python file creating a `paint` module. Copy your classes from above into the module and implement paint functions. Implement a `CompoundShape` class. Create a simple drawing demonstrating that all of your classes are working.

```python
class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for _ in range(height)]  
    def set_pixel(self, row, col, char='*'):
        if 0 <= row < self.height and 0 <= col < self.width:
            self.data[row][col] = char
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))
    def h_line(self, x, y, length, char='*'):
        for i in range(x, x + length):
            self.set_pixel(y, i, char)
    def v_line(self, x, y, length, char='*'):
        for i in range(y, y + length):
            self.set_pixel(i, x, char)
    def line(self, x1, y1, x2, y2, char='*'):
        dx = abs(x2 - x1)
        dy = abs(y2 - y1)
        sx = 1 if x1 < x2 else -1
        sy = 1 if y1 < y2 else -1
        err = dx - dy

        while True:
            self.set_pixel(y1, x1, char)
            if x1 == x2 and y1 == y2:
                break
            e2 = 2 * err
            if e2 > -dy:
                err -= dy
                x1 += sx
            if e2 < dx:
                err += dx
                y1 += sy

class Rectangle:
    def __init__(self, x, y, width, height):
        self.x = x
        self.y = y
        self.width = width
        self.height = height
    def draw(self, canvas):
        canvas.h_line(self.x, self.y, self.width)
        canvas.h_line(self.x, self.y + self.height, self.width)
        canvas.v_line(self.x, self.y, self.height)
        canvas.v_line(self.x + self.width, self.y, self.height)

class Triangle:
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self.points = [(x1, y1), (x2, y2), (x3, y3)]
    def draw(self, canvas):
        (x1, y1), (x2, y2), (x3, y3) = self.points
        canvas.line(x1, y1, x2, y2)
        canvas.line(x2, y2, x3, y3)
        canvas.line(x3, y3, x1, y1)

# Main drawing
if __name__ == "__main__":
    c = Canvas(50, 25)

    # Draw a rectangle and a triangle
    r1 = Rectangle(5, 5, 15, 5)
    t1 = Triangle(20, 20, 35, 5, 45, 20)

    # Draw shapes
    r1.draw(c)
    t1.draw(c)

    c.display()
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for _ in range(height)]
    def set_pixel(self, row, col, char='*'):
        if 0 <= row < self.height and 0 <= col < self.width:
            self.data[row][col] = char
    def v_line(self, x, y, h, **kwargs):
        for i in range(y, y + h):
            self.set_pixel(i, x, kwargs.get('char', '*'))
    def h_line(self, x, y, w, **kwargs):
        for i in range(x, x + w):
            self.set_pixel(y, i, kwargs.get('char', '*'))
    def line(self, x1, y1, x2, y2, **kwargs):
        dx = x2 - x1
        dy = y2 - y1
        steps = max(abs(dx), abs(dy))
        x_inc = dx / steps
        y_inc = dy / steps
        x, y = x1, y1
        for _ in range(steps):
            self.set_pixel(round(y), round(x), kwargs.get('char', '*'))
            x += x_inc
            y += y_inc
    def clear_canvas(self):
        self.data = [[' '] * self.width for _ in range(self.height)]
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class Shape:
    def __init__(self, name="", **kwargs):
        self.name = name
        self.kwargs = kwargs
    def paint(self, canvas):
        pass

class Rectangle(Shape):
    def __init__(self, x, y, w, h, **kwargs):
        Shape.__init__(self, **kwargs)
        self.x = x
        self.y = y
        self.w = w
        self.h = h
    def paint(self, canvas):
        canvas.v_line(self.x, self.y, self.h, **self.kwargs)
        canvas.v_line(self.x + self.w, self.y, self.h, **self.kwargs)
        canvas.h_line(self.x, self.y, self.w, **self.kwargs)
        canvas.h_line(self.x, self.y + self.h, self.w, **self.kwargs)

class Square(Rectangle):
    def __init__(self, x, y, size, **kwargs):
        Rectangle.__init__(self, x, y, size, size, **kwargs)

class Line(Shape):
    def __init__(self, x1, y1, x2, y2, **kwargs):
        Shape.__init__(self, **kwargs)
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2
    def paint(self, canvas):
        canvas.line(self.x1, self.y1, self.x2, self.y2, **self.kwargs)

class CompoundShape(Shape):
    def __init__(self, shapes):
        self.shapes = shapes
    def paint(self, canvas):
        for shape in self.shapes:
            shape.paint(canvas)

class RasterDrawing:
    def __init__(self):
        self.shapes = dict()
        self.shape_names = []
    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()
        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)
    def modify_shape(self, name, new_shape):
        if name in self.shapes:
            self.shapes[name] = new_shape
    def remove_shape(self, name):
        if name in self.shapes:
            self.shape_names.remove(name)
            del self.shapes[name]
    def paint(self, canvas):
        canvas.clear_canvas()
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)
    def assign_name(self):
        name_base = "shape"
        i = 0
        while f"{name_base}_{i}" in self.shapes:
            i += 1
        return f"{name_base}_{i}"

# Testing the Drawing
c1 = Canvas(50, 30)
rd = RasterDrawing()

# Adding shapes
rd.add_shape(Square(5, 5, 10, char="^"))
rd.add_shape(Rectangle(15, 10, 8, 5, char="#"))

# First Paint
print("=== Initial Drawing ===")
rd.paint(c1)
c1.display()

# Modifying shapes: Removing square and adding a line
rd.remove_shape("shape_0")
rd.add_shape(Line(0, 0, 25, 25, char="*"))

# Painting Again
print("\n=== Modified Drawing ===")
rd.paint(c1)
c1.display()
```

12. Implement the ability to load/save raster drawings and demonstate that your method works. One way to implement this ability:

   * Overload `__repr__` functions of all objects to return strings of the python code that would construct the object.
   
   * In the save method of raster drawing class, store the representations into the file.
   * Write a loader function that reads the file and uses `eval` to instantiate the object.

For example:

```python
# Base Shape Class
class Shape:
    def __init__(self, name=""):
        self.name = name
    def paint(self, canvas): pass
    def __repr__(self):
        return "Shape(" + repr(self.name) + ")"


# Rectangle Class
class Rectangle(Shape):
    def __init__(self, x, y, w, h):
        Shape.__init__(self)
        self.x = x
        self.y = y
        self.w = w
        self.h = h
    def paint(self, canvas):
        canvas.v_line(self.x, self.y, self.h)
        canvas.v_line(self.x + self.w, self.y, self.h)
        canvas.h_line(self.x, self.y, self.w)
        canvas.h_line(self.x, self.y + self.h, self.w)
    def __repr__(self):
        return "Rectangle(" + repr(self.x) + "," + repr(self.y) + "," + repr(self.w) + "," + repr(self.h) + ")"


# Square Class
class Square(Rectangle):
    def __init__(self, x, y, size):
        Rectangle.__init__(self, x, y, size, size)
    def __repr__(self):
        return "Square(" + repr(self.x) + "," + repr(self.y) + "," + repr(self.w) + ")"


# RasterDrawing Class
class RasterDrawing:
    def __init__(self):
        self.shapes = dict()
        self.shape_names = list()
    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()
        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)
    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)
    def assign_name(self):
        name_base = "shape"
        name = name_base + "_0"
        i = 1
        while name in self.shapes:
            name = name_base + "_" + str(i)
            i += 1
        return name
    def save(self, filename):
        f = open(filename, "w")
        for sh in self.shapes:
            f.write(self.shapes[sh].__repr__() + "\n")
        f.close()

# Shape Loader Function
def shape_loader(filename):
    f = open(filename, "r")
    tmp = f.read().splitlines()
    f.close()
    return [eval(line) for line in tmp]

# ========== TESTING THE CODE ==========

# Create RasterDrawing instance
rd = RasterDrawing()

# Add shapes
rd.add_shape(Square(5, 5, 10))
rd.add_shape(Rectangle(15, 15, 5, 10))

# Save shapes to a file
print("============saving shapes===========")
rd.save("shapes.txt")
for shape_name in rd.shape_names:
    print(rd.shapes[shape_name])

# Load shapes from the saved file
print("============loading shapes===========")
loaded_shapes = shape_loader("shapes.txt")
for shape in loaded_shapes:
    print(shape)
```
